## Introduction

**Kernel methods** are a powerful class of ML algorithms that work by implicitly mapping data to high-dimensional feature spaces. The key insight is that we never explicitly compute the high-dimensional mapping - we only need the **kernel function** $k(x, x')$ which measures similarity:

$$k(x, x') = \langle \phi(x), \phi(x') \rangle$$

where $\phi(x)$ is the feature map.

**Quantum kernels** replace the classical feature map with a quantum circuit:

$$k_Q(x, x') = |\langle 0|U^\dagger(x) U(x')|0\rangle|^2$$

This measures the overlap of quantum states encoded by $x$ and $x'$, potentially accessing high-dimensional feature spaces that are classically intractable.

## 1) Classical Kernel Methods

Kernel methods work by implicitly mapping data to a high-dimensional feature space where linear separation becomes possible. Common kernel functions include:

1. **Linear Kernel**: $k(x, x') = x \cdot x'$
2. **Polynomial Kernel**: $k(x, x') = (x \cdot x' + c)^d$
3. **RBF (Gaussian) Kernel**: $k(x, x') = \exp(-\gamma ||x - x'||^2)$

The **kernel matrix** (Gram matrix) contains all pairwise similarities:

$$K_{ij} = k(x_i, x_j)$$

Support Vector Machines (SVMs) use the kernel matrix to find the optimal separating hyperplane in the feature space.

## 2) Quantum Kernel Theory

A quantum kernel is computed by encoding two data points into quantum states and measuring their overlap:

$$k_Q(x, x') = |\langle 0|U^\dagger(x) U(x')|0\rangle|^2$$

This can be interpreted as:

1. **State Preparation**: Encode $x$ and $x'$ into quantum states
2. **Overlap Measurement**: Compute the inner product of the states
3. **Kernel Value**: Square of the overlap magnitude

**Key Properties:**
- Quantum kernels are automatically **positive semi-definite**
- They can access **exponentially large** feature spaces
- The encoding circuit $U(x)$ determines the feature map

## 3) Computing the Kernel Matrix

To use a quantum kernel with SVMs, we need to compute the full kernel matrix $K$ where:

$$K_{ij} = k_Q(x_i, x_j)$$

This requires computing the kernel for all pairs of training samples. For $N$ samples, this means $N^2$ kernel evaluations.

**Efficiency Considerations:**
- Each kernel evaluation requires running a quantum circuit
- The kernel matrix must be computed before training the SVM
- For large datasets, consider using subsampling or approximate methods

1. **Parameterized Quantum Kernels**: $k(x, x'; \theta)$ with trainable parameters

3. **Hybrid Kernels**: Combine classical and quantum kernels

**Parameterized Quantum Kernels:**
$$k_Q(x, x'; \theta) = |\langle 0|U^\dagger(x, \theta) U(x', \theta)|0\rangle|^2$$

These can be trained to learn optimal feature maps for specific tasks.
