# Worksheet 1 | Data Encoding in Quantum Machine Learning

In this worksheet, we explore different methods to encode classical data into quantum states. Data encoding is a crucial first step in any quantum machine learning pipeline, as it determines how we represent information in the quantum realm. We will cover:

- **Basis Encoding**: Binary representation using computational basis states
- **Angle Encoding**: Encoding data as rotation angles on qubits
- **Amplitude Encoding**: Encoding data in quantum amplitudes
- **Data Reuploading**: Repeated encoding for enhanced expressibility

## Introduction

In classical machine learning, data is represented as vectors in $\mathbb{R}^n$. In quantum computing, we need to encode this data into **quantum states** (state vectors) in a Hilbert space. The choice of encoding method significantly impacts:

- **Expressibility**: How much of the quantum feature space can be explored?
- **Efficiency**: How many qubits are needed for $n$ features?
- **Compatibility**: How well does it integrate with existing quantum algorithms?

## 1) Basis Encoding

Basis encoding represents data as binary strings, where each bit is encoded as the state of a qubit:

$$|x\rangle = |b_0 b_1 \ldots b_{n-1}\rangle$$

where $b_i \in \{0, 1\}$ and the feature $x$ is represented in binary notation.

**Advantages:**
- Simple to implement
- Uses only $n$ qubits for $n$ bits
- Native representation for binary classification

**Disadvantages:**
- Exponential qubit requirement for $2^n$ possible states
- Sparse state representation (only one basis state is populated)
- Limited expressibility for continuous data

## 2) Quantum Random Access Memory (QRAM)

QRAM allows quantum computers to access classical data stored at multiple memory addresses simultaneously. Unlike classical RAM where you access one address at a time, QRAM can retrieve data from many addresses in superposition:

$$\text{QRAM}: \sum_i \alpha_i |i\rangle_{\text{addr}} \otimes |0\rangle_{\text{data}} \rightarrow \sum_i \alpha_i |i\rangle_{\text{addr}} \otimes |d_i\rangle_{\text{data}}$$

where $d_i$ is the data stored at address $i$, encoded using basis encoding.

**Key Concepts:**
- **Address qubits**: Qubits that store the memory address in superposition
- **Data qubits**: Qubits that store the retrieved data
- **Basis Encoding**: Each data value is stored as a binary pattern using X gates

**Advantages:**
- Parallel access to multiple memory locations
- Enables quantum algorithms that require memory lookups
- Foundation for quantum databases and QML with classical data

**Disadvantages:**
- Requires specialized hardware (bucket brigade or circuit-based)
- Gate complexity scales with memory size
- Decoherence risks with deep circuits

## 3) Angle Encoding

Angle encoding represents data as rotation angles applied to qubits:

$$|x\rangle = \bigotimes_{i=1}^{n} R_Y(\theta_i)|0\rangle = \cos(\theta_i/2)|0\rangle + \sin(\theta_i/2)|1\rangle$$

where $\theta_i = \pi \cdot x_i$ for normalized features $x_i \in [0, 1]$.

**Advantages:**
- Efficient: $n$ features require only $n$ qubits
- Continuous representation of data
- Good expressibility for small feature spaces

**Disadvantages:**
- Limited capacity for high-dimensional data
- Information is encoded in relative phases (harder to extract)

## 4) Amplitude Encoding

Amplitude encoding represents data in the amplitudes of a quantum state:

$$|x\rangle = \sum_{i=1}^{N} x_i |i\rangle$$

where $x_i$ are the normalized features and $N = 2^n$ (number of basis states).

**Key Requirements:**
- Features must be normalized: $\sum_i |x_i|^2 = 1$
- $n$ qubits can encode $2^n$ features

**Advantages:**
- Highly efficient: exponential features in linear qubits
- Good for kernel methods
- Full utilization of quantum state space

**Disadvantages:**
- Requires normalization
- Complex state preparation circuits
- Measurement statistics needed for extraction

From the last two examples you can probably guess, that angle encoding is a kind of amplitude encoding. With angle encoding you directly manipulate amplitudes of each qubit's state through rotation. The final multi-qubit state from angle enconding has amplitudes that are products of the individual qubit amplitudes. Still, both approaches are complementary:

- **Angle encoding** encodes $n$ features into $n$ qubits $\rightarrow$ linear scaling
- **Amplitude encoding** encodes $2^n$ features into $n$ qubits $\rightarrow$ exponential scaling

While angle encoding is simpler to implement and useful for small feature spaces and current NISQ hardware, amplitude encoding achieves higher density but requires more commplex state preparation. Angle encoding can be seen as a *local* amplitude manipulation, whereas amplitude encoding sets amplitudes *globally* across the entire Hilbert space.

## 5) Data Reuploading

The idea of data reuploading stems from the fact how data is used in Neural Networks. It can be accessed in several nodes whereas this is not possible on a quantum circuit except we upload the data several times.

<img src="figs/data_reuploading.png" alt="Source: https://arxiv.org/abs/1907.02085" width="70%"/>

Thus, data reuploading is a technique where the same data is encoded multiple times throughout the circuit:

$$U(x) = W_{l+1}(\theta_{l+1})\left[\prod_{l=1}^{L} S_l(x) \cdot W_l(\theta_l)\right]$$

where $U_l(x)$ re-encodes the data and $W_l$ are variational layers that are paremeterized by parameters $\theta$.

<img src="figs/circuit_data_reuploading.png" width="70%"/>

**Why Data Reuploading?**
- Classical neural networks can learn arbitrary functions through depth
- Quantum circuits need reuploading to achieve similar expressibility
- Single encoding is a linear transformation; reuploading adds complexity

**Advantages:**
- Enhanced expressibility
- Better gradient flow
- Universality guarantees

**Disadvantages:**
- Longer circuits
- More parameters to optimize

## Summary

In this worksheet, we explored four fundamental data encoding methods for quantum machine learning:

1. **Basis Encoding**: Simple binary representation, good for discrete data
2. **Angle Encoding**: Rotation-based, efficient for small feature spaces
3. **Amplitude Encoding**: Highly efficient, exponential features in linear qubits
4. **Data Reuploading**: Enhanced expressibility through repeated encoding

**Key Takeaways:**
- The choice of encoding depends on your data and application
- Amplitude encoding is most efficient for high-dimensional data
- Data reuploading can improve the expressibility of any encoding
- Consider qubit requirements and circuit depth when choosing an encoding
