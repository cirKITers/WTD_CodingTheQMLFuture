    <details>
    <summary>Click here!</summary>
    
  You can normalize your features like this:

  ```python
  norm = np.sqrt(np.sum(np.abs(features)**2))
  if  norm == 0:
    raise ValueError("Input features cannot be all zeros.")
  features_normalized = features / norm
  ```

  Alternatively you can use:

  ```python
  norm = np.linalg.norm(features)
  if norm == 0:
    raise ValueError("Input features cannot be all zeros.")
  features_normalized = features / norm
  ```
  
</details>
</div>

  ```python
  def amplitude_encoding(features, n_qubits):
    """
    Encode features using amplitude encoding.
    """
    norm = np.sqrt(np.sum(np.abs(features)**2))
    if norm == 0:
        raise ValueError("Input features cannot be all zeros.")
    features_normalized = features / norm
    
    @qml.qnode(dev)
    def circuit():
        qml.AmplitudeEmbedding(features_normalized, wires=range(n_qubits), normalize=False)

        return qml.probs()
    
    return circuit
  ```
