  Data reuploading is not limited to repeating the data encoding layerwise. Instead you can also repeat your data by adding more qubits. This has the same effect. So depth can be reduced by increasing width.

  ```python
  def data_reuploading(feature, weights, n_layers):
    """
    Encode feature using data reuploading.
    """
    @qml.qnode(dev)
    def circuit():
        for layer in range(n_layers):
            qml.RX(weights[layer], wires=0)
            qml.RY(feature, wires=0)
        qml.RX(weights[layer+1], wires=0)

        return qml.expval(qml.PauliZ(0)) 
    
    return circuit

  weights = (
    2 * np.pi * np.random.random(size=(n_layers + 1))
  )
  ```
