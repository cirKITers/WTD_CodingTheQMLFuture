  For the variational layer, first apply three rotations. Please check the worksheet on data encodings to pick something appropriate.

  After that follows some entangling gates. For entanglement, you can use the [CNOT gate](https://docs.pennylane.ai/en/stable/code/api/pennylane.CNOT.html). It expects a list of two wires. The first one is the control qubit, the second one the target.

  ```python
  qml.CNOT(wires=[0, 1])
  ```

  When qubit $0$ is $1$, an X gate is applied to qubit $1$.

    <details>
    <summary>Click here!</summary>
    
  For the entanglement, you can either choose a cascade of CNOT gates. This means that all neighboring qubits are connected through a CNOT gate.

  ```python
  for i in range(wires-1):
    qml.CNOT(wires=[i, i+1])
  ```

  Otherwise you can use a closed ring which adds another CNOT gate, that connects the last to the first qubit.

  ```python
  for i in range(wires-1):
    qml.CNOT(wires=[i, i+1])
  qml.CNOT(wires=[wires-1, 0])
  ```

</details>

  def variational_layer(weights, wires):
    """
    Apply a single variational layer.
    """
    for i in range(len(wires)):
        qml.RZ(weights[i, 0], wires=wires[i])
        qml.RY(weights[i, 1], wires=wires[i])
        qml.RZ(weights[i, 2], wires=wires[i])
    # Apply CNOT gates for entanglement
    for i in range(len(wires) - 1):
        qml.CNOT(wires=[wires[i], wires[i + 1]])
    qml.CNOT(wires=[wires[-1], wires[0]])
