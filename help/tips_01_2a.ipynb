    <details>
    <summary>Click here!</summary>
    
  To use Basis Encoding in combination with an address register, it is not sufficient to only put your X gates, but you need to put controlled X gates. This way, the address register gets connected to the data register.

  And in case you are using an address register with more than one qubit, you will need a `MultiControlledX` gate:

  ```python
  qml.MultiControlledX(list(range(n_addr_qubits)) + [data_position])
  ```
  
  The last element in the list is the target where the X operation is applied to.
  
</details>

    <summary>Click here!</summary>
    
  To save a datum to a specific address, first prepare the address register. E.g. in case you want to encode something on address `01` you need to ensure that you put an X gate on the first address qubit. This way you can use a MultiControlledX gate to check for 0 and 1.

  ```python
  def qram_operation(data, n_addr_qubits, n_data_qubits):
    """
    Implement a simplified QRAM operation.
    """
    # Calculate required number of address qubits
    n_addresses = 2 ** n_addr_qubits
    
    # Validate input
    if len(data) != n_addresses:
        raise ValueError(f"Data length ({len(data)}) must match number of addresses ({n_addresses})")
    
    # Create device with address and data qubits
    total_qubits = n_addr_qubits + n_data_qubits
    dev_qram = qml.device("default.qubit", wires=total_qubits)
    
    @qml.qnode(dev_qram)
    def circuit():
        for i in range(n_addr_qubits):
            qml.Hadamard(wires=i)
        
        for address in range(n_addresses):
            for position, bit in enumerate(format(address, f'0{n_addr_qubits}b')):
                if int(bit) == 0:
                    qml.X(position)  # Flip address qubits to match the current address
            # Encode data at the current address
            for position, bit in enumerate(format(data[address], f'0{n_data_qubits}b')):
                if int(bit) == 1:
                    qml.MultiControlledX(list(range(n_addr_qubits)) + [n_addr_qubits + position])
            # Reset address qubits for the next iteration
            for position, bit in enumerate(format(address, f'0{n_addr_qubits}b')):
                if int(bit) == 0:
                    qml.X(position)  # Flip address qubits to match the current address
            
        return qml.state()
    
    return circuit
  ```
