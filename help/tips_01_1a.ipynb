    <summary>Click here!</summary>
    
  The X gate is one of the fundamental quantum gates and corresponds to the classical NOT operation.

  ### Effect

  - $|0\rangle \rightarrow |1\rangle$ (flips the state from 0 to 1)
  - $|1\rangle \rightarrow |0\rangle$ (flips the state from 1 to 0)

  ### Matrix representation

  $
  X = \begin{pmatrix}
  0&1\\
  1&0
  \end{pmatrix}
  $

  
</details>

  ```python
  def basis_encoding(value, n_qubits):
    """
    Encode an integer value using basis encoding.
    """
    assert 0 <= value < 2**n_qubits, "Value must be in range [0, 2^n_qubits - 1]"

    @qml.qnode(dev)
    def circuit():
        for position, bit in enumerate(format(value, f'0{n_qubits}b')):
            if int(bit) == 1:
                qml.X(position)
        return qml.state()
    
    return circuit
  ```
