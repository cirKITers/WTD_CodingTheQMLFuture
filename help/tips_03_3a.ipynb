  <h2><i class="fas fa-life-ring" style="font-size:36px"></i> &nbsp; Tip 1</h2>
    <details>
    <summary>Click here!</summary>
    
  Take care to work with the right set of qubits. After a block of convolution and pooling layers, you should only consider every second wire:

  ```python
  wires = wires[::2]
  ```
  
</details>
</div>

<div class="alert alert-info">
  <h2><i class="fas fa-life-ring" style="font-size:36px"></i> &nbsp; Tip 2</h2>
    <details>
    <summary>Click here!</summary>
    
  You can return the probability for the first wire by using:

  ```python
  qml.probs(wires=(0))
  ```
  
  You will then have a list of two probabilities you can interpret as your different type of class.
</details>

  ```python
  def conv_and_pooling(kernel_weights, wires):

  @qml.qnode(dev)
def qcnn_classifier(conv_weights, fc_weights, features):
    """
    Complete QCNN for binary classification.
    """
    wires = list(range(num_wires))
    layers = conv_weights.shape[1]
    
    qml.AmplitudeEmbedding(features=features, wires=wires, pad_with=0.5)
    qml.Barrier(wires=wires, only_visual=True)

    # adds convolutional and pooling layers
    for j in range(layers):
        conv_and_pooling(conv_weights[:, j], wires)
        wires = wires[::2]
        qml.Barrier(wires=wires, only_visual=True)

    assert fc_weights.size == 4 ** (len(wires)) - 1, (
        "The size of the last layer weights vector is incorrect!"
        f" \n Expected {4 ** (len(wires)) - 1}"
    )
    dense_layer(fc_weights, wires)
    return qml.probs(wires=(0))
  ```
