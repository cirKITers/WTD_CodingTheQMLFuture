## Expressibility and Generalization

A key advantage of QCNNs is their ability to generalize well with limited training data. Research has shown that:

1. **Expressibility**: QCNNs can represent complex functions with fewer parameters than classical CNNs
2. **Generalization**: Quantum circuits can exhibit better generalization on small datasets
3. **Barren Plateaus**: Deep circuits may suffer from vanishing gradients - use shallow circuits for stability. For QCNNs it was proven that there are no barren plateaus.

<img src="../figs/qcnn_generalization.png" width="90%" />

The generalization error can be estimated as:

$$\text{Gen Error} = R(\theta) - \hat{R}_N(\theta)$$

where $R$ is the true risk and $\hat{R}_N$ is the empirical risk on $N$ samples.
